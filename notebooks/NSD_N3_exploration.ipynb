## Summary

- Dataset has 59 sessions across 5 NHP subjects recorded in visual cortex
- Each session has hundreds of units (single and multi-unit activity)
- Neural responses clearly driven by visual stimuli with strong evoked activity (100-250ms)
- Substantial unit-to-unit and image-to-image variability in firing rates
- `response_matrix_img` (units × time × images) is the main analysis array
- Use `evk_fr()` helper for evoked firing rates per unit per image
